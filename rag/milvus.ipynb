# Milvus

```python
from pymilvus import MilvusClient

# =========================
# 初始化Milvus 1. 基本配置
# =========================
MILVUS_URI = "http://localhost:19530" # Milvus 服务的连接地址
DB_NAME = "rag_tutorial" # 自定义数据库名称
COLLECTION_NAME = "docs" # 向量集合名称（类似于传统数据库的表）
KNOWLEDGE_FILE = "../knowledge.txt" # 本地知识库文件路径

# BGE-M3 在 SiliconFlow / Milvus 文档中都是 1024 维
EMBED_MODEL_NAME = "Pro/BAAI/bge-m3" # 嵌入模型名称
EMBED_DIM = 1024 # BGE-M3 模型输出的向量维度固定为 1024
```

```python
# =========================
# 2. 初始化 Milvus
# =========================
# 初始化 Milvus 客户端
client = MilvusClient(MILVUS_URI, "root", "92f28eeee1123cd8ad601d3b32f1516b")

# 创建数据库如果指定的数据库不存在，则主动创建
existing_dbs = client.list_databases()
if DB_NAME not in existing_dbs:
    client.create_database(db_name=DB_NAME)

# 切换到当前工作的数据库
client.use_database(db_name=DB_NAME)

# 创建 collection
client.create_collection(
    collection_name=COLLECTION_NAME,
    dimension=1024, # 向量维度
    metric_type="COSINE" # 使用余弦相似度
)
```

```python
# 查看所有数据库
databases = client.list_databases()
for db in databases:
    print(db)

# 查看数据库下的 collections
collections = client.list_collections()
for collection in collections:
    print(collection)

# 删除数据库
client.drop_database(db_name=DB_NAME)

# 删除 collection
client.drop_collection(collection_name=COLLECTION_NAME)
```
