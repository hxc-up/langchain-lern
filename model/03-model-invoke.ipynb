## 1. invoke 的传参
### 1.1 文本输入

### 1.2 字典列表

```python
import os
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model

load_dotenv(override=True)

DEEPSEEK_API_KEY = os.getenv("DEEPSEEK_API_KEY")
DEEPSEEK_BASE_URL = os.getenv("DEEPSEEK_BASE_URL")

model = init_chat_model(
    model="deepseek:deepseek-flash",
    base_url=DEEPSEEK_BASE_URL,
    api_key=DEEPSEEK_API_KEY,
)

# 创建字典列表组成消息，一条消息通常包含 role（角色）、content（内容）等信息
# 适用于设置系统提示、表达多轮对话历史、因为它JSON兼容，易于序列化和网络传输，所以生产环境推荐

messages1 = [
    {"role": "system", "content": "你是一个专业的数学老师"},
    {"role": "user", "content": "给我解释一下什么是斐波那契数列以及他的作用"},
]
print(model.invoke(messages1))

# 多轮对话
messages2 = [
    {"role": "system", "content": "你是一个专业的数学老师"},
    {"role": "user", "content": "1+7 = ？"},
    {"role": "assistant", "content": "8"},
    {"role": "user", "assistant": "再加3等于多少？"},
]
print(model.invoke(messages2))
```

### 1.3 消息对象列表

```python
from langchain_core.messages import SystemMessage, HumanMessage
import os
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model

load_dotenv(override=True)

DEEPSEEK_API_KEY = os.getenv("DEEPSEEK_API_KEY")
DEEPSEEK_BASE_URL = os.getenv("DEEPSEEK_BASE_URL")

model = init_chat_model(
    model="deepseek:deepseek-flash",
    base_url=DEEPSEEK_BASE_URL,
    api_key=DEEPSEEK_API_KEY,
)

# 针对上面字典列表的方式进一步封装，使用内置消息类（SystemMessage、HumanMessage、AIMessage），将消息对象列表输入模型。
# 适用于需要类型检查（针对大型项目）；缺点：代码较长、不如字典简洁，难以序列化为JSON

messages = [
    SystemMessage(content="你是一个专业的数学老师"),
    HumanMessage(content="给我解释一下什么是斐波那契数列以及他的作用"),
]
print(model.invoke(messages))
```

## 2. invoke 的返回值

```python
from langchain_core.messages import SystemMessage, HumanMessage
import os
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model

load_dotenv(override=True)

DEEPSEEK_API_KEY = os.getenv("DEEPSEEK_API_KEY")
DEEPSEEK_BASE_URL = os.getenv("DEEPSEEK_BASE_URL")

model = init_chat_model(
    model="deepseek:deepseek-flash",
    base_url=DEEPSEEK_BASE_URL,
    api_key=DEEPSEEK_API_KEY,
)

response = model.invoke([HumanMessage(content="2+3*2=?")])
```

```python
print(type(response))
```

```python
from rich import print as rprint

# 美化输出1
response.pretty_print()

# 美化输出2
rprint(response)
```

```python
response = model.invoke("用一句话解释什么是 AI")

# 1.获取回复内容
print("AI 回复:", response)

# 2.获取响应元数据
metadata = response.response_metadata
print(f"使用的模型: {metadata['model_name']}")
print(f"结束原因: {metadata['finish_reason']}")
print(f"模型提供商: {metadata['model_provider']}\n")

# 3.获取 Token 使用情况
usage = metadata.get("token_usage", {})
print(f"输入 tokens: {usage.get('prompt_tokens')}")
print(f"输出 tokens: {usage.get('completion_tokens')}")
print(f"总计 tokens: {usage.get('total_tokens')}\n")

# 4.获取消息 ID
print(f"消息 ID: {response.id}")
```
