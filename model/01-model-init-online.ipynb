# 调用线上的大模型
## 1.调用具体模型厂商的API
### 1.1 调用 deepseek 官网模型


```python
import os

from dotenv import load_dotenv
from langchain_deepseek import ChatDeepSeek

# 读取env配置文件，且以配置文件中的变量优先
load_dotenv(override=True)

# DEEPSEEK_API_KEY = os.getenv("DEEPSEEK_API_KEY")
# DEEPSEEK_BASE_URL = os.getenv("DEEPSEEK_BASE_URL")

llm_deepseek = ChatDeepSeek(
    model="deepseek-flash",
    # api_key=DEEPSEEK_API_KEY,
    # api_base=DEEPSEEK_BASE_URL,
)

response = llm_deepseek.invoke("你是什么模型")
print(response)
```

## 2.兼容的用法
### 1.使用 ChatOpenAI

```python
from langchain_openai import ChatOpenAI

load_dotenv(override=True)

DEEPSEEK_API_KEY = os.getenv("DEEPSEEK_API_KEY")
DEEPSEEK_BASE_URL = os.getenv("DEEPSEEK_BASE_URL")

model = ChatOpenAI(
    model="deepseek-flash",
    base_url=DEEPSEEK_BASE_URL,
    api_key=DEEPSEEK_API_KEY,
)

print(model.invoke("1 + 1 = ?"))
```

## 3.使用 LangChain1.x 的统一方式
### 使用 init_chat_model() 初始化模型

```python
from langchain.chat_models import init_chat_model

load_dotenv(override=True)

DEEPSEEK_API_KEY = os.getenv("DEEPSEEK_API_KEY")
DEEPSEEK_BASE_URL = os.getenv("DEEPSEEK_BASE_URL")

model = init_chat_model(
    # 方式一
    # model="deepseek-flash",
    # model_provider="deepseek",
    # 方式二 PS：如果调用三方平台时，前面的 model_provider 要换成 openai，这个相当于和 ChatOpenAI 方法保持一致
    model="deepseek:deepseek-flash",

    base_url=DEEPSEEK_BASE_URL,
    api_key=DEEPSEEK_API_KEY,
)

print(model.invoke("介绍下你自己"))
```
